```python
import pandas as pd
```

```python
data={
    "customer":["priya","AMIT","rahul","amit"],
    "email":["priya@gmail.com","laya@gmailo.com","rahul@gmail.com","sneha@gmail.com"],
    "phone":["+91-9876643210","9876643211","9876643212","98766432123"],
    "amount":["$1200","1500 INR","Rs.2000","1500"]
}
df=pd.DataFrame(data)
df
```

```python
df["customer"]=df["customer"].str.strip()
df
```

```python
df["customer"]=df["customer"].str.lower()
df
```

```python
df["amount"]=df["amount"].str.replace("Rs." ,"")
df
```

```python
df["amount"]=df["amount"].str.replace(r"[INR.-]","",regex=True)
df
```

```python
df["amount"]=df["amount"].str.replace("$" ,"")
df
```

```python
df["phone"]=df["phone"].str.replace(r"[-+91]","",regex=True)
df
```

```python
df["phone"]="+91"+df["phone"].astype(str)
df
```
