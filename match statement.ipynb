```python
day=3
match day:
    case 1:
        print("mon")
    case 2:
        print("th")
    case 3:
        print("w")
```

```python
day=9
match day:
    case 1|2|3|4|5:
        print("wd")
    case 6|7:
        print("we")
    case _:
        print("invalid")
```

```python
num1=int(input("enter your first value"))
num2=int(input("enter your second value"))

op=input("enter your operator(+,-,/,*)")
print(num1,num2,op)

match op:
    case "+":
        print(num1+num2)
    case "-":
        print(num1-num2)
    case "*":
        print(num1*num2)
    case _:
        print("invalid data")

```
