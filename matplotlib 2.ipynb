```python
days=[1,2,3,4,5]
sales=[100,200,150,300,400]
plt.fill_between(days, sales, color="blue")
plt.title("Area Plot")
plt.show()
```

```python
months=["jan","feb","mar","apr","may","jun"]
electronics=[100,150,165,345,234,432,]
clothing=[78,80,45,70,90,80]
furniture=[50,60,55,65,70,40]

plt.stackplot(
    months,
    electronics,
    clothing,
    furniture,
    labels=[electronics,clothing,furniture,])
plt.xlabel("month")
plt.ylabel("sales(₹thousands)")
plt.title(" monthly sales by product category")

plt.legend()
plt.show()
```

```python
plt.plot(months,furniture)
plt.text(10,20," important point")
plt.show()
```
