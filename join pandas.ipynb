```python
import pandas as pd
```

```python
data={
    "employee":["riya","salu","raju","manu","winu","ammu"],
    "department":["it","hr","it","finance","hr","it"],
    "city":["bangalore","chennai","bangalore","delhi","chennai","mumbai"],
    "salary":[6000,5000,65000,7000,52000,58000],
    "experience":[5,3,7,10,4,6]
}
df = pd.DataFrame(data)
df
```

```python
df.groupby("department")
```

```python
df.groupby("department")["salary"].mean()
```

```python
df.groupby("department")["salary"].sum()
```

```python
df.groupby("department")["salary"].max()
```

```python
df.groupby("department")["salary"].agg(["mean","sum","min","max"])
```

```python
df.groupby("department").agg(
    avg_salary=("salary", "mean"),
    total_salary=("salary","sum"),
    max_experience=("experience","max")
)
```

```python
employees=pd.DataFrame({
    "empid":[1,2,3,4],
    "name":["chinnu", "salu","saabi","bavu"],
    "Dept id":[101,102,101,103],
    "salary":[60000,50000,65000,70000]
})
departments=pd.DataFrame({
    "Dept id":[101,102,104],
    "Department":["it","hr","finance"]
})
employees,departments
```

```python
pd.merge(employees, departments, on="Dept id",how="inner")
```

```python
pd.merge(employees, departments, on="Dept id",how="left")
```

```python
pd.merge(employees, departments, on="Dept id",how="right")
```

```python
pd.merge(employees, departments, on="Dept id",how="outer")
```
