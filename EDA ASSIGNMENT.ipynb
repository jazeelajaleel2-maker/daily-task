```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
sns.countplot(x="Region",data=df)
plt.title("Region wise order count")
plt.show()
```

```python
sns.countplot(x="Category",data=df)
plt.title("Category wise order count")
plt.show()
```

```python
df["Quantity"]=pd.to_numeric(df["Quantity"],errors="coerce")
```

```python
plt.Figure(figsize=(10,5))
pr_sales=df.groupby("Product")["Quantity"].sum().reset_index()
plt.xticks(rotation=45, ha="right")
sns.barplot(data=pr_sales, x="Product",y="Quantity")
plt.title("product wise sold quantity")
plt.xlabel("Product")
plt.ylabel("total quantity sold")
plt.show()
```

```python
sns.countplot(x="Payment_Method", data=df)

plt.title("Payment_Method")
plt.xlabel("Payment_Method")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.Figure(figsize=(10,5))
pr_sales=df.groupby("Sales_Rep")["Quantity"].sum().reset_index()
plt.xticks(rotation=45, ha="right")
sns.barplot(data=pr_sales, x="Sales_Rep",y="Quantity")
plt.title("Sales_Rep wise sale")
plt.xlabel("Sales_Rep")
plt.ylabel("total quantity sold")
plt.show()
```

```python
r_sales=df.groupby("Order_Date")["Quantity"].sum().reset_index()
sns.lineplot(x="Order_Date",y="Quantity",data=df)
plt.title(" Order_Date wise quantity")
plt.xlabel("Order_Date")
plt.ylabel("total quantity sold")
plt.xticks(rotation=45)
plt.show()
```

```python
df[["Quantity", "Unit_Price"]].dtypes
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")
```

```python
df = df.dropna(subset=["Quantity", "Unit_Price"])
```

```python
sns.scatterplot(data=df,x="Quantity",y="Unit_Price")
plt.title("quantity vs Unit_Price")
plt.xlabel("Quantity")
plt.ylabel("Unit_Price")
plt.show()
```

```python
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```

```python
Payment_Method = df["Payment_Method"].value_counts()
```

```python
plt.figure(figsize=(7,7))
plt.pie(Payment_Method, labels=Payment_Method.index,autopct="%1.1f%%")
plt.title("payment method distribution")
plt.show()
```
